```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Parameters
# -----------------------------
np.random.seed(42)
num_steps = 120
step_size = 1

home = np.array([0.0, 0.0]) # Changed to float to prevent UFuncTypeError
position = home.copy()

path_x = [position[0]]
path_y = [position[1]]

# To store movement vectors (path integration)
movement_vectors = []

# -----------------------------
# Exploration Phase
# -----------------------------
for _ in range(num_steps):
    angle = np.random.uniform(0, 2 * np.pi)  # direction w.r.t sun
    move = step_size * np.array([np.cos(angle), np.sin(angle)])

    position += move
    movement_vectors.append(move)

    path_x.append(position[0])
    path_y.append(position[1])

food_location = position.copy()

# -----------------------------
# Return Phase (Path Integration)
# -----------------------------
home_vector = -np.sum(movement_vectors, axis=0)

return_steps = 30
return_path_x = []
return_path_y = []

for i in range(return_steps):
    position += home_vector / return_steps
    return_path_x.append(position[0])
    return_path_y.append(position[1])

# -----------------------------
# Plot
# -----------------------------
plt.figure(figsize=(8, 8))
plt.plot(path_x, path_y, label="Exploration Path", color="blue")
plt.plot(return_path_x, return_path_y, label="Direct Return Path", color="red")
plt.scatter(0, 0, color="green", s=100, label="Home")
plt.scatter(food_location[0], food_location[1], color="orange", s=100, label="Food")

plt.title("Ant-Inspired Path Integration Navigation")
plt.xlabel("X Position")
plt.ylabel("Y Position")
plt.legend()
plt.grid()
plt.show()
# Save the plot
plt.savefig('simulation_result.png', dpi=300, bbox_inches='tight')
plt.close()
```
